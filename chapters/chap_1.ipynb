# Structured Optput
## Langchain provides a structured output format that makes it easy to work with complex data structures and extract specific information from them.
## This format is designed to be easy to parse and process,
## making it ideal for scenarios where you need to extract specific information from complex data structures.
## this makes LLM to interact with other tools like databases,APIs,etc.

## LLm can be of two types
1. LLm that return structured output. Have a function "with_structured_output"
2.LLm that can return unstructured output, We have to parse the output manually.

# So how to use with_structured_output
## we have to call the with_structured_output function before before invokeing our prompt to LLms.
## This function takes a prompt template and returns a new prompt template that can be used to invoke the LLm with structured output.

```
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import GoogleGenerativeAI
import os
load_dotenv()
g_api_key = os.getenv("GEMINI_API_KEY")
if g_api_key is None:
    raise ValueError("Google API key not found in environment variables")
else:
    user_input = input("Please enter your symptoms")
    if user_input == ("exit" or "Exit"):
        exit
    else:
        llm_gemini = GoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0.2,
                                        api_key=g_api_key)

        sample_prmpt_template = ChatPromptTemplate.from_messages(
            [
                ("system", "You are a top notch doctor with experince of 35 years."),
                ("user", "{user_input}"),

            ])
        structured_model=llm_gemini.with_structured_output(
            sample_prmpt_template
        )
        result= structured_model.invoke()
```

# Data Format can be done ia many ways
## 1. typedDict - it is a python data structure that is used to define the structure of a data object. It is a dictionary where the keys are the names of the fields and the values are the types of the fields. And it only allows to access the keys that are defined in the dictionary.
## 2. Pydantic
## 3. Json_Schema

```
from typing import TypedDict
class Symptoms(TypedDict):
    symptoms: str
    symptoms_list: list[str]
    age:int
patient = Symptoms(
    symptoms="Fever",
    symptoms_list=["Fever","Cough","Sore throat"],
    age=29
)
print(patient)
```

```
from typing import TypedDict, Annotated
class test_output_structure(TypedDict):
    review: str
    model :str
    score: int
    sentiment: Annotated[str,"The sentiment of the review."]
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
import os
load_dotenv()
g_api_key = os.getenv("GEMINI_API_KEY")
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite",
                           temperature=0.2,
                           api_key=g_api_key)
structured_model =model.with_structured_output(
    test_output_structure
)
result = structured_model.invoke(" i have brought an xiomi phone model 5Aseries, it has bad camera and battery life is low. I would gave 3 stars to the model")
print(result)
```

# Paralell Chains

```
from langchain_google_genai import GoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
import os
from dotenv import load_dotenv
load_dotenv()
gemini_key = os.getenv("GEMINI_API_KEY")
grok_key = os.getenv("GROK_API_KEY")

#checking for keys
if (gemini_key or grok_key) is None :
    print("Key not found")
else:
    print("WE have the keys, please move ahead")

google_powered_llm = GoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    api_key = gemini_key,
    temperature= 0.2
)
grok_powered_llm = ChatGroq(
    model = "llama3-8b-8192",
    temperature = 0.2,
    api_key = grok_key
)

# Prompts
prompt_1 = PromptTemplate(
    template = "Use given text to generate summarised notes of {text}",
    input_variables = ["text"]
)
prompt_2 = PromptTemplate(
    template = " Generate questions and answers for the given text {text}",
    input_variables = ["text"]
)
prompt_3 = PromptTemplate(
    template = "Merge both the {Summary} and set of {question_answers} into a single document",
    input_variables = ["Summary","question_answers"]
)
parser = StrOutputParser()
# Now Preparing Chains

paralell_chain = RunnableParallel(
    {"Summary": prompt_1 | google_powered_llm | parser,
    "question_answers": prompt_2 | google_powered_llm | parser}
)
merge_chain = RunnableParallel(
    {"merged_text": paralell_chain | prompt_3 | google_powered_llm | parser}
)
text = """LangChain is a framework for creating applications using large language models (LLMs). It connects LLMs like GPT-4 or Gemini with the needs of developers. LangChain uses a suite of tools. These tools allow developers to combine prompt templates, models, and output parsers into workflows.One of LangChain's key features is "Chains." Chains allow the output of one LLM to be the input for another. This creates multi-step processes. LangChain also manages memory. This lets AI agents remember past interactions and keep context in a conversation. "Retrieval Augmented Generation" (RAG) lets models access private documents or external databases. This helps provide accurate answers.The framework includes components for document loaders, text splitters, and vector store integrations. These components make it easier to build searchable knowledge bases. LangChain supports Python and JavaScript. This broadens its use for software engineers. The "Agents" feature lets LLMs decide which tools to use. These tools can include a calculator or a web search. This modularity allows developers to switch models easily.Tools like LangServe and LangSmith help with deployment and debugging. This helps ensure applications are ready for production. LangChain is an architecture for intelligent software. It allows builders to create systems that can reason, search, and act. It lowers the barrier to entry for AI development. LangChain provides the tools needed to build a summarizer or a complex assistant.The LangChain community is active. It adds new integrations for the latest AI research and tools. This keeps the framework current. Its ability to handle structured data, unstructured text, and code makes it versatile. LangChain is a toolkit for turning model capabilities into applications. It allows developers to focus on innovation. The framework's interoperability supports various vector databases, LLM providers, and APIs. LangChain’s monitoring and evaluation tools help maintain output quality. It bridges the gap between prototypes and production systems. LangChain connects the components of the modern AI stack. This enables the creation of digital experiences.
"""

result = merge_chain.invoke({
    "text": text
})
print(result)
```

# Conditional Chains


```
from langchain_google_genai import GoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnableParallel,RunnableBranch,RunnableLambda
from langchain_core.output_parsers import PydanticOutputParser, StrOutputParser
from typing import Annotated, Literal
from pydantic import BaseModel,Field
import os
load_dotenv()

gemini_key = os.getenv("GEMINI_API_KEY")
grok_key = os.getenv("GROK_API_KEY")

#checking for keys
if (gemini_key or grok_key) is None :
    print("Key not found")
else:
    print("WE have the keys, please move ahead")

google_powered_llm = GoogleGenerativeAI(
    model="gemini-2.5-flash-lite",
    api_key = gemini_key,
    temperature= 0.2
)
grok_powered_llm = ChatGroq(
    model = "llama3-8b-8192",
    temperature = 0.2,
    api_key = grok_key
)

prompt = PromptTemplate(
    template = " Classify the given feedback into either positive or negative category {feedback}. And also the message that we can reply with \n {format_instruction}",
    input_variables = ["feedback"],
    partial_variables = {"format_instruction":parser.get_format_instructions()}
)
prompt_to_handle_negative_feedback = PromptTemplate(
    template = " Generate a response to handle the given negative feedback, {feedback}",
    input_variables = ["feedback"]
)


prompt_to_handle_positive_feedback = PromptTemplate(
    template = " Generate a response to handle the given positive feedback, {feedback}",
    input_variables = ["feedback"]
)

# pydantic class to get output from llm in desired output format

class pydantic_class(BaseModel):
    category: Literal["positive","negative"]

parser_x = PydanticOutputParser(pydantic_object=pydantic_class)
parser_y = StrOutputParser()

chain = prompt | google_powered_llm | parser_x
#chain_2 =prompt | google_powered_llm | parser_y


#result_2 = chain_2.invoke({
#    "feedback":"I am very dissatisfied with the product"
#})

branch_chain = RunnableBranch(
    (lambda x: x.category == "positive",  prompt_to_handle_positive_feedback | google_powered_llm | parser_y),
    (lambda x:x.category == "negative", prompt_to_handle_negative_feedback | google_powered_llm | parser_y),
    RunnableLambda(lambda x: "I am not sure what to do with the feedback")
)
merged_chain = chain | branch_chain
result = merged_chain.invoke({
    "feedback":"I am very satisfied with the product"
})
print(result)
```
