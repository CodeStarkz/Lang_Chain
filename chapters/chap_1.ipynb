# Structured Optput
## Langchain provides a structured output format that makes it easy to work with complex data structures and extract specific information from them.
## This format is designed to be easy to parse and process,
## making it ideal for scenarios where you need to extract specific information from complex data structures.
## this makes LLM to interact with other tools like databases,APIs,etc.

## LLm can be of two types
1. LLm that return structured output. Have a function "with_structured_output"
2.LLm that can return unstructured output, We have to parse the output manually.

# So how to use with_structured_output
## we have to call the with_structured_output function before before invokeing our prompt to LLms.
## This function takes a prompt template and returns a new prompt template that can be used to invoke the LLm with structured output.

```
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import GoogleGenerativeAI
import os
load_dotenv()
g_api_key = os.getenv("GEMINI_API_KEY")
if g_api_key is None:
    raise ValueError("Google API key not found in environment variables")
else:
    user_input = input("Please enter your symptoms")
    if user_input == ("exit" or "Exit"):
        exit
    else:
        llm_gemini = GoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0.2,
                                        api_key=g_api_key)

        sample_prmpt_template = ChatPromptTemplate.from_messages(
            [
                ("system", "You are a top notch doctor with experince of 35 years."),
                ("user", "{user_input}"),

            ])
        structured_model=llm_gemini.with_structured_output(
            sample_prmpt_template
        )
        result= structured_model.invoke()
```

```
# The BOt
from langchain_google_genai import GoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel,Field
from typing import Annotated
import os
load_dotenv()
g_api_key = os.getenv("GEMINI_API_KEY")
grk_api_key= os.getenv("GROK_API_KEY")

#Check for missing api keys
if grk_api_key is None or g_api_key is None:
    raise ValueError("Please set the GEMINI_API_KEY and grok_api_key in the environment variables")
else:
    model_grk = ChatGroq(model="llama3-8b-8192", temperature=0.2, api_key=grk_api_key)
    model_gemini = GoogleGenerativeAI(
        model="gemini-2.5-flash-lite",
    temperature=0.2,
    api_key=g_api_key
    )
    print("Both model up and running")

# Defining Parser class
class Parser_class_to_achive(BaseModel):
    plan:str
    reference:list[str]

#Template
inital_prompt_template = PromptTemplate(
    template = "I want to master {topic_to_learn}, and currently I am good at {topic_i_know}",
    input_variables = ["topic_to_learn","topic_i_know"],
    partial_variables = {"topic_i_know":parser.get_format_instructions()}
)
final_prompt_template = PromptTemplate(
     template = "Prepare a holistic plan to achive mentioned goal {goal} ",
     input_variables = ["goal"],
    partial_variables = {"goal":parser.get_format_instructions()}
 )

#Declaring Parser
parser = PydanticOutputParser(pydantic_object=Parser_class_to_achive)

prompt_i = inital_prompt_template.format(topic_to_learn="System Design",topic_i_know="python and OOPs")
prompt_f = final_prompt_template.format(goal="to master system design")

chain = prompt_i | model_grk  | parser | prompt_f | model_gemini| parser

result = chain.invoke()
print(result)
```

```
from langchain_google_genai import GoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from pydantic import BaseModel, Field
import os
from dotenv import load_dotenv

load_dotenv()

# 1. Define Parser First
class Parser_class_to_achive(BaseModel):
    plan: str = Field(description="The learning path")
    reference: list[str] = Field(description="List of resources")

parser = PydanticOutputParser(pydantic_object=Parser_class_to_achive)

# 2. Setup Models
model_grk = ChatGroq(model="llama3-8b-8192", temperature=0.2)
model_gemini = GoogleGenerativeAI(model="gemini-1.5-flash", temperature=0.2)

# 3. Refined Templates
# We include {format_instructions} as a separate variable
inital_prompt_template = PromptTemplate(
    template = "I want to master {topic_to_learn}, and currently I am good at {topic_i_know}.\n{format_instructions}",
    input_variables = ["topic_to_learn", "topic_i_know"],
    partial_variables = {"format_instructions": parser.get_format_instructions()}
)

final_prompt_template = PromptTemplate(
     template = "Prepare a holistic plan to achieve this goal: {goal}.\n{format_instructions}",
     input_variables = ["goal"],
     partial_variables = {"format_instructions": parser.get_format_instructions()}
)

# 4. Construct the Chain
# We use a lambda to extract the 'plan' from the first model's output to use as 'goal' for the second
chain = (
    inital_prompt_template


    | model_grk
    | parser
    | (lambda x: {"goal": x.plan})  # Bridges the two prompts


    | final_prompt_template
    | model_gemini
    | parser
)

# 5. Invoke
result = chain.invoke({"topic_to_learn": "System Design", "topic_i_know": "Python and OOPs"})
print(result.plan)
print(result.reference)
```
